```python
# -*- coding: utf-8 -*-
"""
项目依赖库版本检测脚本
运行后输出当前Python环境中所有依赖的版本号
"""

import sys
print(f"Python 版本: {sys.version.split()[0]}")
print("-" * 45)

# 导入名 : 显示名
lib_list = [
    ("numpy",      "NumPy"),
    ("pandas",     "Pandas"),
    ("scipy",      "SciPy"),
    ("sklearn",    "Scikit-learn"),
    ("imblearn",   "Imbalanced-learn"),
    ("xgboost",    "XGBoost"),
    ("lightgbm",   "LightGBM"),
    ("shap",       "SHAP"),
    ("matplotlib", "Matplotlib"),
    ("seaborn",    "Seaborn"),
    ("openpyxl",   "OpenPyXL"),
]

for import_name, display_name in lib_list:
    try:
        module = __import__(import_name)
        print(f"{display_name:18s}  {module.__version__}")
    except ImportError:
        print(f"{display_name:18s}  未安装")

print("-" * 45)
print("检测完成。缺失依赖可执行：pip install -r requirements.txt")
```
